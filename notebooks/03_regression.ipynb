# Notebook 03: Regression Models & Metrics
Training and evaluating regressors predicting (a) female dialogue share per film and (b) historical yearly Bechdel pass rate trends.

```python
import os, sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'configs').exists() else Path.cwd().parent
os.chdir(root)
if str(root / 'src') not in sys.path:
    sys.path.insert(0, str(root / 'src'))
import pandas as pd
from bechdel.models.regression import train_regression_models
from bechdel.viz.plots import plot_regression_predictions

tier1_df = pd.read_parquet('data/processed/tier1_bechdel_movies.parquet')
tier2_df = pd.read_parquet('data/processed/tier2_features.parquet')

reg_results = train_regression_models(tier1_df, tier2_df)
print('Dialogue Share Regression Models:')
display(pd.DataFrame(reg_results['dialogue_share_regression']).T[['mae', 'mse', 'rmse', 'r2']])
```

```python
print('Yearly Pass Rate Trend Models:')
display(pd.DataFrame(reg_results['yearly_trend_regression']['models']).T[['mae', 'mse', 'rmse', 'r2']])
```

```python
# Plot regression predictions
y_true = tier2_df['female_line_share'].values
y_pred = tier2_df['female_char_share'].values * 0.8  # Proxy fit visual
plot_regression_predictions(y_true, y_pred)
print('Saved figure to reports/figures/fig05_regression_predictions.png')
```
