# Notebook 02: Exploratory Data Analysis & Statistical Analysis
Statistical hypothesis testing, correlation matrices, and historical trend analysis. All inferences strictly represent observational correlations and do not imply causation.

```python
import os, sys
from pathlib import Path
root = Path.cwd() if (Path.cwd() / 'configs').exists() else Path.cwd().parent
os.chdir(root)
if str(root / 'src') not in sys.path:
    sys.path.insert(0, str(root / 'src'))
import pandas as pd
import numpy as np
from bechdel.eval.statistical_tests import run_all_statistical_tests
from bechdel.viz.plots import plot_yearly_pass_rate_trend, plot_genre_representation, plot_correlation_matrix

tier1_df = pd.read_parquet('data/processed/tier1_bechdel_movies.parquet')
if 'decade' not in tier1_df.columns:
    tier1_df['decade'] = (tier1_df['year'] // 10) * 10
tier2_df = pd.read_parquet('data/processed/tier2_features.parquet')
print(f'Tier 1: {len(tier1_df)} | Tier 2: {len(tier2_df)}')
```

```python
# 1. Statistical Hypothesis Tests (Strict Non-Causal Framing)
stats_bundle = run_all_statistical_tests(tier1_df, tier2_df)
for test_name, res in stats_bundle.items():
    print(f'=== {test_name} ===')
    for k, v in res.items():
        if k != 'disclaimer' and k != 'pass_rates_by_group':
            print(f'  {k}: {v}')
    print(f'  Disclaimer: {res.get("disclaimer", "")}\n')
```

```python
# 2. Historical Representation Trend Visualization
plot_yearly_pass_rate_trend(tier1_df)
plot_genre_representation(tier2_df)
plot_correlation_matrix(tier2_df)
print('Figures generated in reports/figures/')
```
